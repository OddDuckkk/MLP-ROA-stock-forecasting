# Data Initialization


This study uses three different stocks for stock price forecasting: Bank Central Asia (BBCA), NVIDIA (NVDA), and Tesla (TSLA). These stocks represent different markets and industries, providing diverse time-series data for evaluating the forecasting models.
- BBCA: Indonesian banking sector (IDX).
- NVDA: United States technology sector (NASDAQ).
- TSLA: United States automotive and technology sector (NASDAQ).

Historical daily stock data from the last five years (2019-02-23 - 2024-02-23) will be used for each stock. The datasets will be initialized, filtered, and sorted chronologically to prepare them for preprocessing, model training, and forecasting.

Data Source: https://www.kaggle.com/datasets/tiwill/saham-idx?resource=download (LQ45)
https://www.kaggle.com/datasets/parsalatifi/s-and-p-500-us-stock-market-data?resource=download (S&P500)

```python
import os
import glob
import math
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Load BCA stock data
df_bca = pd.read_csv("Data/Master Data/forecastingsaham/Saham/LQ45/BBCA.csv")

# Convert the Date column to datetime format
df_bca["date"] = pd.to_datetime(df_bca["date"])

# Sort data by ticker and date
df_bca = df_bca.sort_values(
    by=["date"]
).reset_index(drop=True)

# Get the latest available date in the dataset
latest_date = df_bca["date"].max()

# Calculate the starting date for the last five years
start_date = latest_date - pd.DateOffset(years=5)

# Filter data within the last five years
df_bca_5years = df_bca[
    df_bca["date"] >= start_date
].copy()

# Save stock dataset to CSV files
df_bca_5years.to_csv("Data/BBCA_5Years.csv", index=False)

# Display dataset information
print("Date range:", start_date.date(), "-", latest_date.date())

print("\nBBCA records:", len(df_bca_5years))

print("\nBBCA preview:")
print(df_bca_5years.head())
```

```python
# Load US Master Stock Data
df_us = pd.read_csv("Data/Master Data/master_stock_data.csv")

# Filter NVDA and TSLA stocks
df_us_filtered = df_us[df_us["Ticker"].isin(["NVDA", "TSLA"])].copy()

# Convert the Date column to datetime format
df_us_filtered["Date"] = pd.to_datetime(df_us_filtered["Date"])

# Sort data by ticker and date
df_us_filtered = df_us_filtered.sort_values(
    by=["Ticker", "Date"]
).reset_index(drop=True)

# Get the latest available date in the dataset
latest_date = pd.Timestamp("2024-02-23") #to match BCA's latest available date

# Calculate the starting date for the last five years
start_date = latest_date - pd.DateOffset(years=5)

# Filter data within the last five years
df_5years = df_us_filtered[
    (df_us_filtered["Date"] >= start_date) &
    (df_us_filtered["Date"] <= latest_date)
].copy()

# Separate NVDA and TSLA datasets
df_nvda = df_5years[df_5years["Ticker"] == "NVDA"].copy()
df_tsla = df_5years[df_5years["Ticker"] == "TSLA"].copy()

# Rename stock feature columns to match BBCA dataset
rename_columns = {
    "Date": "date",
    "Close": "close",
    "Open": "open_price",
    "High": "high",
    "Low": "low",
    "Volume": "volume"
}

# Rename columns
df_nvda = df_nvda.rename(columns=rename_columns)
df_tsla = df_tsla.rename(columns=rename_columns)

# Save each stock dataset to separate CSV files
df_nvda.to_csv("Data/NVDA_5Years.csv", index=False)
df_tsla.to_csv("Data/TSLA_5Years.csv", index=False)

# Display dataset information
print("Date range:", start_date.date(), "-", latest_date.date())

print("\nNVDA records:", len(df_nvda))
print("TSLA records:", len(df_tsla))

print("\nNVDA preview:")
print(df_nvda.head())

print("\nTSLA preview:")
print(df_tsla.head())
```
